```python
# Remove spaces from the beginning and end of column names
df.columns = df.columns.str.strip()

# Display the cleaned column names
print(df.columns.tolist())
```

```python
# Store all duplicate records
duplicate_rows = df[df.duplicated(keep="first")]

# Count duplicates belonging to each attack class
print("Duplicate records by attack class:")
print(duplicate_rows["Attack type"].value_counts())
```

```python
print("Total records:", len(df))
print("\nAttack class distribution:")
print(df["Attack type"].value_counts())
```

```python
# Create a cleaned copy without exact duplicate rows
df_clean = df.drop_duplicates().copy()

# Display the new dataset size
print("Original dataset size:", df.shape)
print("Cleaned dataset size:", df_clean.shape)

# Calculate the number of removed records
print("Duplicates removed:", len(df) - len(df_clean))
```

```python
# Check whether any duplicates remain
print("Remaining duplicates:", df_clean.duplicated().sum())

# Check attack class distribution
print("\nAttack class distribution after cleaning:")
print(df_clean["Attack type"].value_counts())

# Check dataset information
print("\nDataset information:")
df_clean.info()
```

```python
# Separate input features
X = df_clean.drop("Attack type", axis=1)

# Separate target labels
y = df_clean["Attack type"]

# Display their dimensions
print("Features shape:", X.shape)
print("Target shape:", y.shape)

# Display target classes
print("\nTarget classes:")
print(y.unique())
```

```python
from sklearn.preprocessing import LabelEncoder

# Create a label encoder
label_encoder = LabelEncoder()

# Convert attack names into numerical labels
y_encoded = label_encoder.fit_transform(y)

# Display the mapping
for number, attack in enumerate(label_encoder.classes_):
    print(attack, "=", number)
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

# Display dataset sizes
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
import pandas as pd

print("Training class distribution:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nTesting class distribution:")
print(pd.Series(y_test).value_counts().sort_index())
```

```python
print("Total training records:", len(X_train))

print("Unique ID values:", X_train["id"].nunique())

print("Minimum ID:", X_train["id"].min())

print("Maximum ID:", X_train["id"].max())
```

```python
X_train.describe().T
```

```python
# Remove the ID column
X_train_model = X_train.drop(columns=["id"])
X_test_model = X_test.drop(columns=["id"])

# Check the updated dimensions
print("Training data:", X_train_model.shape)
print("Testing data:", X_test_model.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Learn scaling values from training data and transform it
X_train_scaled = scaler.fit_transform(X_train_model)

# Use the same scaling values for testing data
X_test_scaled = scaler.transform(X_test_model)

# Display dimensions
print("Scaled training data:", X_train_scaled.shape)
print("Scaled testing data:", X_test_scaled.shape)
```

```python
import numpy as np

print("Mean of first 5 scaled features:")
print(np.mean(X_train_scaled, axis=0)[:5])

print("\nStandard deviation of first 5 scaled features:")
print(np.std(X_train_scaled, axis=0)[:5])
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean",
    n_jobs=-1
)
```

```python
knn_model.fit(X_train_scaled, y_train)

print("KNN model training completed!")
```

```python
y_pred = knn_model.predict(X_test_scaled)

print("Prediction completed!")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("KNN Accuracy:", accuracy * 100, "%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

fig, ax = plt.subplots(figsize=(9, 7))
display.plot(ax=ax, cmap="Blues", values_format="d")
plt.title("KNN Confusion Matrix")
plt.xticks(rotation=45)
plt.show()
```

```python
train_ids = set(X_train["id"])
test_ids = set(X_test["id"])

common_ids = train_ids.intersection(test_ids)

print("Unique training node IDs:", len(train_ids))
print("Unique testing node IDs:", len(test_ids))
print("Node IDs appearing in both sets:", len(common_ids))
```

```python
from sklearn.model_selection import GroupShuffleSplit
# Use the dataset without the ID feature
X_group = X.drop(columns=["id"])

# Use the sensor IDs only for grouping
groups = df_clean["id"]

# Create a group-based 80:20 split
group_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    group_split.split(X_group, y_encoded, groups=groups)
)

# Create training and testing datasets
X_group_train = X_group.iloc[train_idx]
X_group_test = X_group.iloc[test_idx]

y_group_train = y_encoded[train_idx]
y_group_test = y_encoded[test_idx]

print("Group training shape:", X_group_train.shape)
print("Group testing shape:", X_group_test.shape)

train_node_ids = set(groups.iloc[train_idx])
test_node_ids = set(groups.iloc[test_idx])

common_node_ids = train_node_ids.intersection(test_node_ids)

print("Training nodes:", len(train_node_ids))
print("Testing nodes:", len(test_node_ids))
print("Common nodes:", len(common_node_ids))
```

```python
import pandas as pd

print("Group training distribution:")
print(pd.Series(y_group_train).value_counts().sort_index())

print("\nGroup testing distribution:")
print(pd.Series(y_group_test).value_counts().sort_index())
```

```python
from sklearn.preprocessing import StandardScaler

# Create a new scaler
group_scaler = StandardScaler()

# Fit using only group training data
X_group_train_scaled = group_scaler.fit_transform(X_group_train)

# Apply the same scaling to group testing data
X_group_test_scaled = group_scaler.transform(X_group_test)

print("Group training scaled:", X_group_train_scaled.shape)
print("Group testing scaled:", X_group_test_scaled.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier

group_knn = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean",
    n_jobs=-1
)

group_knn.fit(X_group_train_scaled, y_group_train)

print("Group-wise KNN training completed!")
```

```python
y_group_pred = group_knn.predict(X_group_test_scaled)

print("Group-wise prediction completed!")
print("Total predictions:", len(y_group_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

group_accuracy = accuracy_score(
    y_group_test,
    y_group_pred
)

print("Node-wise KNN Accuracy:", group_accuracy * 100, "%")

print("\nClassification Report:")

print(classification_report(
    y_group_test,
    y_group_pred,
    labels=[0, 1, 2, 3, 4],
    target_names=label_encoder.classes_,
    zero_division=0
))
```

```python
import os
import joblib

# Create a folder to store prepared datasets
os.makedirs("../dataset/processed", exist_ok=True)

# Save all prepared data
prepared_data = {
    "X_group_train_scaled": X_group_train_scaled,
    "X_group_test_scaled": X_group_test_scaled,
    "y_group_train": y_group_train,
    "y_group_test": y_group_test,
    "label_encoder": label_encoder,
    "group_scaler": group_scaler
}

joblib.dump(
    prepared_data,
    "../dataset/processed/prepared_data.pkl"
)

print("Prepared datasets saved successfully!")
```
